**Number Pattern Semi Pyramid**

```python
#my answer
step_number=7
for j in range(step_number):
    for i in range(step_number):
        if i<=j:
            print(i+1, end=" ")
    print("")
```

**Exclaimation Pyramid Pattern**

```python
#my answer
n=5
for row in range(n):
    for column in range(n):
        if column==0:
            print("!", end=" ")
        else:
            print("*", end=" ")
    print("")
```

**Pyramid Pattern**

```python
#my answer
n=5
max_num_of_stars=(n*2)-1
for row in range(1, n+1):
    spaces=n-row
    num_of_stars=max_num_of_stars-(spaces*2)
    for column in range(1, max_num_of_stars+1):
        if column<=spaces or column>(num_of_stars+spaces):
            print(" ", end="")
        else:
            print("*", end="")
    print("")
```

**Hollow Rectangle**

```python
#my answer
n=5
for row in range(1, n+1):
    for column in range(1, n+1):
        if column==1 or column==5 or row==1 or row==5:
            print("*", end="")
        else:
            print(" ", end="")
    print("")
```

**Water Image Star Pattern**

```python
#my answer
n=5
for row in range(n):
    spaces=n-(row+1)
    for column in range(n):
        if column<spaces:
            print(" ", end="")
        else:
            print("*", end="")
    print("")
```

**Water Image Number Pattern**

```python
#my answer
n=5
for row in range(n):
    spaces=n-(row+1)
    inverted_nums=row
    for column in range(n):
        if column<spaces:
            print(" ", end="")
        else:
            print(inverted_nums, end="")
            inverted_nums-=1
    print("")
```

**Right Half Diamond Pattern**

```python
#my answer
n=5
total_rows=(n*2)-1
for row in range(total_rows):
    spaces=n-(row+1)
    if row>=n:
        spaces=(row+1)-n
    for column in range(n):
        if column<n-spaces:
            print("*", end="")
        else:
            print(" ", end="")
    print("")
```

**Mirrored Half Diamond Star Patter**

```python
#my answer
n=5
total_rows=(n*2)-1
for row in range(total_rows):
    spaces=n-(row+1)
    if row>=n:
        spaces=(row+1)-n
    for column in range(1, n+1):
        if column>spaces:
            print("*", end="")
        else:
            print(" ", end="")
    print("")
```

**Tricky Numbered Pattern - 2**

```python
#my answer
n=5
new_row_start_num=0
row_sum=1
number_counter=1
for row in range(0, n):
    if row==0:
        print(number_counter)
        continue
    elif row%2==0:
        new_row_start_num=row_sum/row
    else:
        new_row_start_num=(row_sum/row)+1
    number_counter=new_row_start_num
    row_sum=0
    for column in range(0, row+1):
        print(int(number_counter), end=" ")
        row_sum+=number_counter
        number_counter+=2 
    print("")
```

**Diamond Pattern - 1**

```python
#my answer
n=5
epicentre=-(n//-2)
for row in range(1,n+1):
    spaces=(epicentre-row)*2
    if row>=epicentre:
        spaces=(row-epicentre)*2
    num_of_stars=n-spaces
    for column in range(1, n+1):
        if column<=spaces/2 or column>num_of_stars+(spaces/2):
            print(" ", end="")
        else:
            print("*", end="")
    print("")
```

**Diamond Pattern - 2**

```python
#my answer
n=5
max_col=(n*2)-1
counter=0 #for spaces in between the stars
for row in range(1, max_col+1):
    printed_stars=0
    spaces=max_col-row
    if row>n:
        spaces=row-1
    num_of_stars=max_col-spaces
    for column in range(1, max_col+1):
        if counter==1:
            print(" ", end="")
            counter-=1
            continue
        elif column<=n-num_of_stars or num_of_stars==printed_stars:
            print(" ", end="")
        else:
            print("*", end="")
            printed_stars+=1
            counter+=1
    print("")
```

**Hollow Full Pyramid**

```python
#my answer
n=6
spaces=n
last_row_counter=1 #for last row spaces
for row in range(1, n+1):
    spaces-=1
    num_of_stars=n-spaces
    for column in range(1, n*2):
        if row==n and column!=last_row_counter:
            print(" ", end="")
            last_row_counter+=2
        elif column<=spaces or column>spaces+((num_of_stars*2)-1) or (column>spaces+1 and column<spaces+((num_of_stars*2)-1) and row!=n):
            print(" ", end="")
        else:
            print("*", end="")
    print("")
```

**Palindrome Number Pyramid**

```python
#my answer
n=6
for row in range(1, n+1):
    max_counter=0
    for column in range(1, row*2):
        if column>row:
            max_counter-=1
            print(max_counter, end=" ")
        else:
            max_counter+=1
            print(max_counter, end=" ")
    print("")
```
